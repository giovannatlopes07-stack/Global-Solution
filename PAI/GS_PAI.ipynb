```python
import ollama
import random

def rodar_tomada_de_decisao(temperatura, energia, comunicacao):
    """Aplica estruturas lógicas locais para tomada de decisão básica e alertas."""
    alertas = []
    acoes_automatizadas = []

    # Lógica para Temperatura
    if temperatura > 90:
        alertas.append("ALERTA CRÍTICO: Superaquecimento dos módulos!")
        acoes_automatizadas.append("AÇÃO: Ativando sistema de resfriamento líquido emergencial.")
    elif temperatura < 0:
        alertas.append("AVISO: Temperatura abaixo do ideal.")
        acoes_automatizadas.append("AÇÃO: Ligando aquecedores internos.")

    # Lógica para Energia
    if energia < 20:
        alertas.append("ALERTA CRÍTICO: Bateria em nível crítico!")
        acoes_automatizadas.append("AÇÃO: Desativando sistemas secundários. Modo de Economia de Energia ATIVADO.")
    elif energia > 95:
        acoes_automatizadas.append("NOTA: Bateria totalmente carregada. Desconectando excedente solar.")

    # Lógica para Comunicação
    if comunicacao == "instável" or comunicacao == "desconectado":
        alertas.append("ALERTA: Instabilidade severa na telemetria espacial.")
        acoes_automatizadas.append("AÇÃO: Rotacionando antena parabólica principal para reenquadramento de sinal.")

    return alertas, acoes_automatizadas

def obter_analise_ia(temperatura, energia, comunicacao, alertas, acoes):
    """Envia o relatório consolidado para o Llama analisar e prever riscos."""

    contexto_sistema = (
        "Você é o Mission Control AI, o sistema de inteligência artificial de controle de uma missão espacial experimental. "
        "Sua função é analisar os dados de telemetria recebidos, validar as ações automáticas tomadas pelo sistema local, "
        "gerar previsões de riscos futuros e fornecer insights acionáveis de forma clara, técnica e concisa."
    )

    prompt_usuario = f"""
    --- TELEMETRIA DA MISSÃO ---
    Temperatura dos Módulos: {temperatura}°C
    Nível de Energia: {energia}%
    Status de Comunicação: {comunicacao}

    Alertas Locais Disparados: {', '.join(alertas) if alertas else 'Nenhum'}
    Ações de Contingência Executadas: {', '.join(acoes) if acoes else 'Nenhuma'}
    ----------------------------
    Avalie o status atual da missão e forneça uma breve análise com previsões ou recomendações.
    """

    resposta = ollama.chat(
        model="llama3.2:1b",
        messages=[
            {"role": "system", "content": contexto_sistema},
            {"role": "user", "content": prompt_usuario}
        ]
    )
    return resposta["message"]["content"]

def exibir_painel_controle(temperatura, energia, comunicacao):
    """Gera o output visual formatado do painel de controle da missão."""
    # Processamento local de regras
    alertas, acoes = rodar_tomada_de_decisao(temperatura, energia, comunicacao)

    # Processamento Cognitivo via IA
    analise_ia = obter_analise_ia(temperatura, energia, comunicacao, alertas, acoes)

    # Renderização da Interface Visual (Terminal/Colab)
    print("=" * 60)
    print("                MISSION CONTROL STATUS ATUAL                ")
    print("=" * 60)
    print(f" Telemetria Base:")
    print(f"   - Temperatura dos Módulos : {temperatura}°C")
    print(f"   - Nível de Energia Solar  : {energia}%")
    print(f"   - Status de Comunicação   : {comunicacao.upper()}")
    print("-" * 60)

    print(" Painel de Alertas e Decisões Automatizadas:")
    if alertas:
        for alerta in alertas:
            print(f"   [!] {alerta}")
        for acao in acoes:
            print(f"   [*] {acao}")
    else:
        print("   [✔] Todos os sistemas operando dentro dos parâmetros normais.")
    print("-" * 60)

    print(" Análise Cognitiva & Previsões da IA:")
    print(analise_ia)
    print("=" * 60)
    print("\n")

# --- SIMULAÇÃO DOS CENÁRIOS DA MISSÃO ESPACIAL ---

# Cenário 1: Tudo Operando em Perfeito Estado (Dados Normais)
print("[PRODUZINDO CENÁRIO 1: OPERAÇÃO NOMINAL]")
exibir_painel_controle(temperatura=24, energia=85, comunicacao="estável")

# Cenário 2: Situação Crítica de Superaquecimento e Queda de Energia (Dados Críticos)
print("[PRODUZINDO CENÁRIO 2: FALHA OPERACIONAL CRÍTICA]")
exibir_painel_controle(temperatura=95, energia=14, comunicacao="instável")
```
